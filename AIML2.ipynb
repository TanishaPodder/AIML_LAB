## Problem 2: The Water Jug Problem (Solved using BFS)

**Problem Statement:**
You are given two jugs: Jug A with a capacity of 4 litres, and Jug B with a capacity of 3 litres. Neither jug has any markings on it. Using only the following operations — (i) fill a jug completely, (ii) empty a jug completely, (iii) pour water from one jug into the other until either the first jug is empty or the second jug is full — determine the minimum sequence of steps needed to measure out exactly 2 litres of water in any one jug.

```python
from collections import deque

def water_jug_problem(capacity_a, capacity_b, target):
    """
    Solves the Water Jug Problem using Breadth-First Search (BFS).

    Args:
        capacity_a (int): The capacity of Jug A.
        capacity_b (int): The capacity of Jug B.
        target (int): The target amount of water to measure.

    Returns:
        list: A list of tuples representing the sequence of states from start to target,
              or None if the target is unreachable.
    """
    queue = deque()
    visited = set()
    parent = {}

    initial_state = (0, 0)
    queue.append((initial_state, [])) # (current_state, path_to_current_state)
    visited.add(initial_state)
    parent[initial_state] = None

    while queue:
        (jug_a, jug_b), path = queue.popleft()

        if jug_a == target or jug_b == target:
            # Reconstruct path
            final_path = []
            current = (jug_a, jug_b)
            while current is not None:
                final_path.append(current)
                current = parent.get(current)
            return final_path[::-1] # Reverse to get path from start to end

        # Define possible operations and their resulting states
        possible_next_states = []

        # 1. Fill Jug A
        possible_next_states.append((capacity_a, jug_b))
        # 2. Fill Jug B
        possible_next_states.append((jug_a, capacity_b))
        # 3. Empty Jug A
        possible_next_states.append((0, jug_b))
        # 4. Empty Jug B
        possible_next_states.append((jug_a, 0))

        # 5. Pour A to B
        # Amount to pour is min of water in A or space in B
        pour_a_to_b = min(jug_a, capacity_b - jug_b)
        possible_next_states.append((jug_a - pour_a_to_b, jug_b + pour_a_to_b))

        # 6. Pour B to A
        # Amount to pour is min of water in B or space in A
        pour_b_to_a = min(jug_b, capacity_a - jug_a)
        possible_next_states.append((jug_a + pour_b_to_a, jug_b - pour_b_to_a))

        for next_state in possible_next_states:
            if next_state not in visited:
                visited.add(next_state)
                parent[next_state] = (jug_a, jug_b)
                queue.append((next_state, path + [next_state]))

    return None # Target not reachable
```

```python
# Define jug capacities and target amount
JUG_A_CAPACITY = 4
JUG_B_CAPACITY = 3
TARGET_AMOUNT = 2

print(f"Solving Water Jug Problem: Jug A = {JUG_A_CAPACITY}L, Jug B = {JUG_B_CAPACITY}L, Target = {TARGET_AMOUNT}L")

solution_path = water_jug_problem(JUG_A_CAPACITY, JUG_B_CAPACITY, TARGET_AMOUNT)

if solution_path:
    print(f"Minimum steps to get {TARGET_AMOUNT}L:")
    for i, (a, b) in enumerate(solution_path):
        print(f"Step {i}: Jug A: {a}L, Jug B: {b}L")
else:
    print(f"Target of {TARGET_AMOUNT}L is not reachable with given jug capacities.")
```
